```python
import pandas as pd
```

```python
def calculate_cost(
    df: pd.DataFrame,
    price_input: float,
    price_cache: float,
    price_output: float,
) -> float:
    uncached_input = (
        df["input_tokens"].sum()
        - df["cached_input_tokens"].sum()
    )

    return (
        uncached_input * price_input
        + df["cached_input_tokens"].sum() * price_cache
        + df["output_tokens"].sum() * price_output
    ) / 1_000_000
```

```python
nano_medium_detect_flex = pd.read_csv('../data/results/usage/binary/question_detection_gpt-5-4-nano_medium_flex_usage.csv')
mini_high_extract_flex = pd.read_csv('../data/results/usage/extracted_question/relevant_question_extraction_gpt-5-4-mini_high_flex_usage.csv')
nano_medium_detect_flex.shape, mini_high_extract_flex.shape
```

```python
nano_cost_flex = calculate_cost(nano_medium_detect_flex, price_input=0.10, price_cache=0.01, price_output=0.625)
mini_cost_flex = calculate_cost(mini_high_extract_flex, price_input=0.375, price_cache=0.0375, price_output=2.25)
```

```python
print(f'Custo do modelo GPT-5-4 Nano para a tarefa de detecção de perguntas: ${nano_cost_flex:.2f}')
print(f'Custo do modelo GPT-5-4 Mini para a tarefa de extração de perguntas relevantes: ${mini_cost_flex:.2f}')
print(f'Custo total: ${nano_cost_flex + mini_cost_flex:.2f}')
```
